```python
from langchain_core.prompts import ChatPromptTemplate

sys_prompt = ChatPromptTemplate.from_messages([
    ("system", 
     """You are are a helpful for debugging code assistenat.
     You need to find out from the code:
     1. what is the code performing on the scale of 1 to 10, where 1 is very bad and 10 is excellent.
     2. share the technical depth of the code.
     3. give me the critical issues with the code.
     4. what are the pros and cons of the code.
     5. give me the code confidence scrore from 0 to 1 based on the code quality and the code performance.
     6. give me the reason of the confidence score.
     7 suggstions to improve the code.
     """),
     ("human", "Analyze the following code and provide your assessment:{code}")
])
```

```python
# lets fix the structure o/p

from pydantic import BaseModel, Field
from typing import List

class CodeAnalysis(BaseModel):
    code_performance: int = Field(..., ge=1, le=10)
    technical_depth: str
    critical_issues: List[str]
    pros_and_cons: List[str]
    confidence_score: float = Field(..., ge=0, le=1)
    confidence_reason: str
    improvement_suggestions: str
```
