### complex chain from chain1 -> chain2

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_openai import AzureChatOpenAI
```

```python
# def the story propmt to craft a story around a topic
story_prompt = PromptTemplate.from_template("Craft a story about the {topic}")
```

```python
# def output parser to display
output_parser = StrOutputParser()
```

```python
story_chain = story_prompt | chat_llm | output_parser
```

```python
result = story_chain.invoke({"topic": "tom and jerry"})
```

```python
print(result)
```

```python
# def analytical propmt for find the mood inside the story
analytical_propmt = PromptTemplate.from_template("Find the mood of the {story}")
```

```python
analytical_chain = analytical_propmt | chat_llm | output_parser
```

```python
# combine both chain
enhance_chain = RunnablePassthrough.assign(
    story = story_chain
).assign(
    analysis = analytical_chain
)
```

```python
enhance_result = enhance_chain.invoke({"topic": "a rainy day"})
```

```python
print("\nOriginal topic:", enhance_result['topic'])
print("\nGenerated Story:", enhance_result['story'])
print("\nMood Analysis:", enhance_result['analysis'])
```

### Using LangGraph - a stateful workflows and routing


While LCEL is excellent for linear sequences, production systems often require branching logic, cycles, and state management
. LangGraph is the recommended framework for this, allowing you to define workflows as graphs consisting of "nodes" (Python functions) and "edges" (connections between functions)
.
Instead of just passing variables forward, LangGraph uses a master "State" (typically a Python dictionary) that each node reads from and updates

```python
from langchain_core.messages import SystemMessage, HumanMessage
from typing_extensions import TypedDict
from typing import Literal
from langgraph.graph import StateGraph, START, END
```

```python
# openai key def
az_aoai_key = "place_your_key"
az_model="palce_your_model_name"
az_endpoint="place_your_endpoint"
api_ver="place_api_version"

chat_llm = AzureChatOpenAI(
    api_key=az_aoai_key,
    api_version=api_ver,
    model=az_model,
    azure_endpoint=az_endpoint,
    temperature=0.01
)
```

```python
# 1. def workflow's state schema

class JobApplicationState(TypedDict):
    job_description: str
    is_suitable: bool
    application: str


# 2-A. def nodes - analysis job description
def analyze_job_description(state):
    print("Analyzing job decription with LLM...")
    job_description = state["job_description"]
    system_prompt = """
    you are a career advisor. Analyze the job description below and decide if the job is suitable for a candidate with 3+ years of
    python and AI/ML experience.
    Answer with only "Yes" or "No".
    """
    human_prompt = f"Job Description {job_description}"
    #call llm to find the suitability
    response = chat_llm.invoke(
        [
            SystemMessage(content=system_prompt),
            HumanMessage(content=human_prompt)
        ]
    )
    # parse output to check the suitability
    is_suitable = response.content.strip().upper() == "YES"
    return {"is_suitable": is_suitable}


# 2-B. def generate the application
def generate_job_application(state):
    print("Generating the cover letter...")
    job_description = state["job_description"]
    system_prompt = """
    You are an editor, craft a suitable cover letter for the job application within 100 words limit.
    """
    human_prompt = f"Job Application {job_description}"
    response = chat_llm.invoke([
        SystemMessage(content=system_prompt),
        HumanMessage(content=human_prompt)
    ])
    cover_letter = response.content
    return {"application": cover_letter}

# 3. def conditional edge- routing logic
def is_suitable_condition(state: JobApplicationState) -> Literal["generate", END]:
    if state.get("is_suitable"):
        return "generate"
    return END

# 4. build and compile the complete workflow graph
builder = StateGraph(JobApplicationState)

# adding nodes
builder.add_node("analyze", analyze_job_description)
builder.add_node("generate", generate_job_application)

# adding edge
builder.add_edge(START, "analyze")
builder.add_conditional_edges("analyze", is_suitable_condition)
builder.add_edge("generate", END)

# grapth
graph = builder.compile()

# 5. run the graph
final_state = graph.invoke({"job_description": "Senior Python Developer needed."})
print(final_state)
```
